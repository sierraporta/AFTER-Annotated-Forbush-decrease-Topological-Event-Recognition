```python
from pathlib import Path
import pandas as pd

base = Path("Results_v2_1/")

def read_csv_safe(name):
    path = base / name
    if not path.exists():
        print(f"[MISSING] {name}")
        return None
    df = pd.read_csv(path)
    print(f"\n=== {name} ===")
    print(df.head())
    print(f"shape = {df.shape}")
    return df

# 1) Cobertura
cov = read_csv_safe("station_coverage_report.csv")
if cov is not None:
    print("\n--- station coverage summary ---")
    cols = [c for c in ["counts_missing_pct", "bg_missing_pct", "deltaN_missing_pct", "A_complex_missing_pct", "n_spikes_flagged"] if c in cov.columns]
    print(cov[["station"] + cols].to_string(index=False))

# 2) Catálogos por nivel
for level in ["strict", "medium", "loose"]:
    cat = read_csv_safe(f"AFTER_catalog_{level}.csv")
    if cat is not None:
        print(f"\n--- {level} catalog summary ---")
        if "catalog_tier" in cat.columns:
            print("\nTier counts:")
            print(cat["catalog_tier"].value_counts(dropna=False))
        if "quality" in cat.columns:
            print("\nQuality counts:")
            print(cat["quality"].value_counts(dropna=False))
        if "year" in cat.columns:
            print("\nCounts by year:")
            print(cat["year"].value_counts().sort_index())
        if {"catalog_tier", "year"}.issubset(cat.columns):
            print("\n2019 by tier:")
            print(cat.loc[cat["year"] == 2019, "catalog_tier"].value_counts(dropna=False))
        if {"n_stations", "drop_mean"}.issubset(cat.columns):
            print("\nKey descriptive stats:")
            print(cat[["n_stations", "drop_mean"]].describe())

# 3) Validación
for level in ["strict", "medium", "loose"]:
    val = read_csv_safe(f"validation_summary_{level}.csv")
    if val is not None:
        print(f"\n--- {level} validation summary ---")
        keep = [c for c in ["subset", "tolerance_h", "n_izmiran", "n_catalog", "n_matched", "recall", "precision", "f1", "median_abs_dt_h"] if c in val.columns]
        print(val[keep].to_string(index=False))

# 4) Matches más importantes
for fname in [
    "match_table_medium_core_18h.csv",
    "match_table_medium_all_18h.csv",
    "match_table_strict_core_18h.csv"
]:
    df = read_csv_safe(fname)
    if df is not None:
        print(f"\n--- {fname} important columns ---")
        cols = [c for c in ["dt_hours", "t_izmiran", "repr_time", "n_stations", "quality", "catalog_tier", "drop_mean"] if c in df.columns]
        print(df[cols].head(15).to_string(index=False))
```

## **Proving re-calibration**

```python
from pathlib import Path
import numpy as np
import pandas as pd

base = Path("Results_v2_2")

cat = pd.read_csv(base / "AFTER_catalog_detector.csv")
izm = pd.read_csv(base / "izmiran_2019_parsed.csv")

cat["repr_time"] = pd.to_datetime(cat["repr_time"], utc=True, errors="coerce")
izm["t_izmiran"] = pd.to_datetime(izm["t_izmiran"], utc=True, errors="coerce")

cat2019 = cat.loc[cat["repr_time"].dt.year == 2019].copy()
izm2019 = izm.loc[izm["t_izmiran"].dt.year == 2019].copy()

def one_to_one_time_match(ref_times, cand_times, tolerance_hours=18.0):
    tol = pd.Timedelta(hours=tolerance_hours)
    pairs = []
    for i, t_ref in enumerate(ref_times):
        for j, t_cand in enumerate(cand_times):
            dt = t_cand - t_ref
            if abs(dt) <= tol:
                pairs.append((i, j, dt))
    if not pairs:
        return pd.DataFrame(columns=["ref_pos", "cand_pos", "dt_hours"])
    pairs = sorted(pairs, key=lambda x: abs(x[2].total_seconds()))
    used_ref = set()
    used_cand = set()
    kept = []
    for i, j, dt in pairs:
        if i in used_ref or j in used_cand:
            continue
        used_ref.add(i)
        used_cand.add(j)
        kept.append({
            "ref_pos": i,
            "cand_pos": j,
            "dt_hours": dt.total_seconds() / 3600.0
        })
    return pd.DataFrame(kept)

def evaluate_subset(subset_df, izm_df, tolerances=(12, 18, 24)):
    ref_times = list(izm_df["t_izmiran"])
    cand_times = list(subset_df["repr_time"])
    rows = []
    for tol in tolerances:
        m = one_to_one_time_match(ref_times, cand_times, tolerance_hours=tol)
        n_ref = len(ref_times)
        n_cand = len(cand_times)
        n_match = len(m)
        recall = n_match / n_ref if n_ref else np.nan
        precision = n_match / n_cand if n_cand else np.nan
        f1 = (2 * precision * recall / (precision + recall)) if np.isfinite(precision) and np.isfinite(recall) and (precision + recall) > 0 else np.nan
        rows.append({
            "tolerance_h": tol,
            "n_catalog": n_cand,
            "n_matched": n_match,
            "recall": recall,
            "precision": precision,
            "f1": f1,
            "median_abs_dt_h": m["dt_hours"].abs().median() if not m.empty else np.nan,
        })
    return pd.DataFrame(rows)

# -------------------------------------------------
# A) Solo por número de estaciones soportantes
# -------------------------------------------------
support_thresholds = [2, 3, 4, 5, 6]

rows = []
detail_tables = {}

for k in support_thresholds:
    sub = cat2019.loc[cat2019["n_supported_stations"] >= k].copy()
    ev = evaluate_subset(sub, izm2019, tolerances=(12, 18, 24))
    ev["rule"] = f"n_supported_stations >= {k}"
    ev["n2019"] = len(sub)
    ev["quality_counts"] = str(sub["quality"].value_counts().to_dict())
    rows.append(ev)
    detail_tables[k] = sub

summary_k = pd.concat(rows, ignore_index=True)
print("\n=== Calibration by n_supported_stations ===")
print(summary_k.to_string(index=False))

# Mejor por F1 a 18h
best18 = summary_k.loc[summary_k["tolerance_h"] == 18].sort_values("f1", ascending=False)
print("\n=== Best by F1 at 18h ===")
print(best18.to_string(index=False))

# -------------------------------------------------
# B) Número de estaciones + amplitud media de soporte
# -------------------------------------------------
drop_thresholds = [2.5, 3.0, 3.5, 4.0]

rows2 = []

for k in [2, 3, 4]:
    for d in drop_thresholds:
        sub = cat2019.loc[
            (cat2019["n_supported_stations"] >= k) &
            (cat2019["mean_supported_drop"] >= d)
        ].copy()
        ev = evaluate_subset(sub, izm2019, tolerances=(12, 18, 24))
        ev["rule"] = f"n_supported_stations >= {k} & mean_supported_drop >= {d}"
        ev["n2019"] = len(sub)
        rows2.append(ev)

summary_kd = pd.concat(rows2, ignore_index=True)
print("\n=== Calibration by support count + mean supported drop ===")
print(summary_kd.to_string(index=False))

best18_kd = summary_kd.loc[summary_kd["tolerance_h"] == 18].sort_values("f1", ascending=False)
print("\n=== Best combined rules by F1 at 18h ===")
print(best18_kd.head(15).to_string(index=False))

# -------------------------------------------------
# C) Variante más estricta: exigir al menos una estación de soporte extra
# -------------------------------------------------
rows3 = []

for k in [2, 3, 4]:
    sub = cat2019.loc[
        (cat2019["n_supported_stations"] >= k) &
        (cat2019["n_support_only_stations"] >= 1)
    ].copy()
    ev = evaluate_subset(sub, izm2019, tolerances=(12, 18, 24))
    ev["rule"] = f"n_supported_stations >= {k} & n_support_only_stations >= 1"
    ev["n2019"] = len(sub)
    rows3.append(ev)

summary_support_only = pd.concat(rows3, ignore_index=True)
print("\n=== Calibration requiring at least one support-only station ===")
print(summary_support_only.to_string(index=False))

best18_support_only = summary_support_only.loc[summary_support_only["tolerance_h"] == 18].sort_values("f1", ascending=False)
print("\n=== Best support-only rules by F1 at 18h ===")
print(best18_support_only.to_string(index=False))
```
